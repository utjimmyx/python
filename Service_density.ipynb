```python
# A customer service representative must spend different amounts of time with each customer to resolve various concerns. The amount of time spent with each customer can be modeled by the following distribution: X ~ Exp (0.2).A customer service representative must spend different amounts of time with each customer to resolve various concerns. The amount of time spent with each customer can be modeled by the following distribution: X ~ Exp (0.2).
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import expon

# -----------------------------
# Parameters
# -----------------------------
rate = 0.2
p = 0.70

# 70th percentile
x_70 = expon.ppf(p, scale=1/rate)

# x values
x = np.linspace(0, 20, 500)

# PDF and CDF
pdf = expon.pdf(x, scale=1/rate)
cdf = expon.cdf(x, scale=1/rate)

# -----------------------------
# Create two plots
# -----------------------------
plt.figure(figsize=(12, 5))

# ===== PDF =====
plt.subplot(1, 2, 1)

plt.plot(x, pdf)

# Shade the area below the 70th percentile
x_shade = np.linspace(0, x_70, 200)
y_shade = expon.pdf(x_shade, scale=1/rate)

plt.fill_between(x_shade, y_shade, alpha=0.3)

# 70th percentile line
plt.axvline(x_70, linestyle='--',
            label=f'70th percentile = {x_70:.2f}')

plt.xlabel('Time (minutes)')
plt.ylabel('Probability Density')
plt.title('PDF: Probability Density Function')
plt.legend()
plt.grid(alpha=0.3)


# ===== CDF =====
plt.subplot(1, 2, 2)

plt.plot(x, cdf)

# 70% horizontal line
plt.axhline(0.70, linestyle='--',
            label='70% probability')

# 70th percentile vertical line
plt.axvline(x_70, linestyle='--',
            label=f'70th percentile = {x_70:.2f}')

plt.xlabel('Time (minutes)')
plt.ylabel('Cumulative Probability')
plt.title('CDF: Cumulative Distribution Function')
plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
print(f"70th percentile: {x_70:.2f} minutes")
print(f"PDF at 70th percentile: {expon.pdf(x_70, scale=1/rate):.4f}")
print(f"CDF at 70th percentile: {expon.cdf(x_70, scale=1/rate):.2f}")
```
